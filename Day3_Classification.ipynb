```python
#Understanding missing values and filling them
data.isnull().sum()
```

```python
#fill the missing values in Age with mean
data['Age'] = data['Age'].fillna(data['Age'].mean(),inplace=True)
data['Age'].isnull().sum()
```

```python
data.isnull().sum()
```

```python
data['Embarked'] = data['Embarked'].fillna('S', inplace=True)
```

```python
data.isnull().sum()
```

```python
#Feature Engineering ---->Feature Encoding(process of convert categorical values into numerical)
data.columns
```

```python
#get_dummies()
pclass = pd.get_dummies(data['Pclass'], drop_first=True, dtype=int)
pclass
```

```python
gender = pd.get_dummies(data ['Sex'], dtype=int)
gender
```

```python
embarked = pd.get_dummies(data ['Embarked'], drop_first=True, dtype=int)
gender
```

```python
#Concatination of all tables
final_data = pd.concat([data, pclass, gender, embarked], axis=1)
final_data
```

```python
final_data.columns
```

```python
final_data.drop(columns=['PassengerId','Name',
                         'Pclass','Ticket',
                         'Sex','Embarked'], inplace=True)
final_data.head()
```

```python
#Logistic Regression --->Binary Classification approach(Survived -0/1)
#Apply modeling process

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, StratifiedKFold, StratifiedShuffleSplit, cross_val_score
from sklearn.metrics import(
        accuracy_score, precision_score, recall_score,f1_score,roc_auc_score,
        confusion_matrix, classification_report, roc_curve, ConfusionMatrixDisplay)
```

```python
#Loading the data
model_data = final_data.copy()
#model_data
#model_data.info()
```

```python
model_data.columns = model_data.columns.astype(str)
X = model_data.drop(columns=['Survived','Cabin'])
y = model_data['Survived'].astype(int)
X.info()
y
```

```python
print("Input Shape:",X.shape)
print("Output Shape:",y.shape)
```

```python
#Holdout Method --> splitting the data & Crossvalidation
X_train,X_test,y_train,y_test = train_test_split(X,y,
                                                test_size=0.2,
                                                stratify=y,
                                                random_state=42)
logistic_model = Pipeline([
    ("scale",StandardScaler()),
    ("classifier",LogisticRegression(max_iter=1000,
                                     random_state=42))])
logistic_model
```

```python
#start building the model
logistic_model.fit(X_train,y_train)
logistic_model_prob = logistic_model.predict_proba(X_test)[:,1]
logistic_predictions = logistic_model.predict(X_test)
print(logistic_model_prob[:10])
print(logistic_predictions[:10])
```

```python
print(confusion_matrix(y_test, logistic_predictions))
```

```python
(98+46) /179
#Precision = TP/TP+FP
#Recall = TP/TP+FN
```

```python
print(classification_report(y_test, logistic_predictions))
```

```python
#create an evaluate function
def evaluate_model(name, y_true, prediction, probability):
    """Evaluate Function for Metrics"""
    tn,fp,fn,tp = confusion_matrix(y_true, prediction, labels=[0,1]).ravel()
    return{
    'Model_Name':name,
    'Accuracy':accuracy_score(y_true, prediction),
    'Precision':precision_score(y_true, prediction,zero_division=0),
    'Recall': recall_score(y_true,prediction,zero_division=0),
    'Specificity': tn/ (tn+fp),
    'F1-Score': f1_score(y_true,prediction,zero_division=0),
    'RDC-AUC': roc_auc_score(y_true, probability),
    'TN': tn, "FP": fp, "TP": tp, "FN":fn,}
logistic_metrics = evaluate_model("Logistic Regression",
                                  y_test,
                                  logistic_predictions,
                                  logistic_model_prob)
logistic_metrics
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, logistic_predictions,cmap='Blues',
                                        display_labels=['Didnot Survive', 'Survived'])
plt.title = ("Confusion Matrix")
plt.show()
```

```python
#Check for best models by performing validation 
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
models = {
    "Logistic Regression" : logistic_model,
    "Decision Tree" : DecisionTreeClassifier(max_depth=5,
                                             min_samples_leaf=1,random_state=42),
    "Random Forest" : RandomForestClassifier(n_estimators=150,
                                             min_samples_leaf=3,random_state=42, n_jobs=-1)}
test_outputs = {}
metric_rows =[]
for name,model in models.items():
    model.fit(X_train,y_train)
    probability = model.predict_proba(X_test)[:,1]
    predictions = model.predict(X_test)
    test_outputs[name] = {'probability' : probability,
                          'Prediction' : predictions}
    metric_rows.append(evaluate_model(name,y_test,predictions,probability))
holdout_metrics = pd.DataFrame(metric_rows)
holdout_metrics  
```
